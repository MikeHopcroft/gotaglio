# Reasoning Sample with MCP Server

This example demonstrates using a reasoning model with an an-process MCP server.
Our MCP server, which is implemented in [mcp_server.py](mcp_server.py), provides the following two tools:
* **stage1(text: str) -> str:** returns `text[::-1]`. This is the characters of `text` in reverse order.
* **stage2(text: str) -> str:** returns `text.upper()`. This is the characters of `text` converted to upper case.

We prompt the reasoning model to return `stage2(stage1(user_text))`, using the following prompt:
~~~txt
    Your goal is to transform the user input by passing it to stage1.
    and then passing the result to stage2. Return the result of stage2.
~~~
